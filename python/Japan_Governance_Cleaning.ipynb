```python
## Ministry of Finance Data Extraction

## Three metrics were extracted from 20 annual MoF survey files (FY2005-FY2024):

## Capital Expenditure (Sheet 6 for FY2005-FY2014, Sheet 7 for FY2015-FY2024).
  ## All Industries Total, excluding Finance and Insurance. Unit: 100 million yen.

## Cash Dividends (Sheet 4 across all files).
  ## Total corporate dividends paid, excluding Finance and Insurance. Unit: 100 million yen.

## Earned Surplus (Sheet 5, available FY2015-FY2024 only).
 ## Accumulated retained corporate profits, All Industries. Unit: 100 million yen.

## Each file contains 5 years of data with overlapping coverage across consecutive files.
## Where duplicate years exist the most recently published value will be retained.
## Sheet numbering was changed between FY2014 and FY2015.

# Scanning all MoF annual files and report sheet structures.
import os
import glob
import pandas as pd
from openpyxl import load_workbook
import warnings

mof_folder = "data/raw/mof_annual/"
mof_files = sorted(glob.glob(mof_folder + "*.xlsx"))

print(f"Total files found: {len(mof_files)}")
print()

for filepath in mof_files:
    filename = os.path.basename(filepath)
    try:
        # Try using pandas ExcelFile.
        xl = pd.ExcelFile(filepath)
        print(f"{filename}: {xl.sheet_names}")
    except ValueError as e:
        print(f"Error with {filename}: {str(e)}")
        try:
            # Alternative approach using openpyxl.
            warnings.simplefilter("ignore")
            wb = load_workbook(filepath, read_only=True, data_only=True)
            print(f"{filename} (using openpyxl): {wb.sheetnames}")
        except Exception as e2:
            print(f"Could not open {filename} with openpyxl either: {str(e2)}")
```

```python
# Checking year coverage across all files.
import glob

mof_folder = "data/raw/mof_annual/"
all_files = sorted(glob.glob(mof_folder + "*.xlsx"))

# Performing sheet mapping.
def get_capex_sheet(filename):
    year = int(''.join(filter(str.isdigit, filename))[:4])
    if 2005 <= year <= 2014:
        return '6'
    elif year >= 2015:
        return '7'
    else:
        return None

all_years_covered = set()

for filepath in all_files:
    filename = os.path.basename(filepath)
    sheet = get_capex_sheet(filename)
    if sheet is None:
        print(f"{filename}: Skipped - early format")
        continue
    
    try:
        df = pd.read_excel(filepath, sheet_name=sheet, header=None)
        years_found = []
        for i in range(min(10, len(df))):
            row = df.iloc[i]
            years = [int(v) for v in row 
                    if isinstance(v, (int, float)) 
                    and 1998 <= v <= 2025
                    and v == int(v)]
            if len(years) >= 2:
                years_found = years
                all_years_covered.update(years)
                break
        print(f"{filename} (Sheet {sheet}): {years_found}")
    except Exception as e:
        print(f"{filename}: ERROR — {e}")

print(f"\nAll years covered: {sorted(all_years_covered)}")
print(f"\nMissing years from 2002-2024: {[y for y in range(2002, 2025) if y not in all_years_covered]}")
```

```python
# MoF Capital Expenditure Extraction.
# Extracting All Industries Total CapEx from FY2005-FY2024 files.

def extract_capex_from_file(filepath, sheet_name):
    """
    Extracts year and All Industries capex values from a MoF annual file.
    Returns a dictionary of {year: capex_value} in 100 million yen.
    """
    df = pd.read_excel(filepath, sheet_name=sheet_name, header=None)
    
    results = {}
    year_row_idx = None
    years = {}
    
    # Finding rows containing year numbers.
    for i in range(min(15, len(df))):
        row = df.iloc[i]
        year_cols = {}
        for j, val in enumerate(row):
            if isinstance(val, (int, float)) and 1998 <= val <= 2025 and val == int(val):
                year_cols[j] = int(val)
        if len(year_cols) >= 2:
            year_row_idx = i
            years = year_cols
            break
    
    if not years:
        return {}
    
    # Finding the All Industries row, searching within 20 rows of year row.
    for i in range(year_row_idx, min(year_row_idx + 20, len(df))):
        cell = df.iloc[i, 0]
        if isinstance(cell, str) and ('All Industries' in cell or 
                                       'Total' in cell):
            # Extracting values from year columns.
            for col, year in years.items():
                val = df.iloc[i, col]
                if isinstance(val, (int, float)) and not pd.isna(val) and val > 0:
                    results[year] = val
            if results:
                break
    
    return results

# Processing all files.
mof_folder = "data/raw/mof_annual/"
all_files = sorted(glob.glob(mof_folder + "*.xlsx"))

# Storing all extracted data with source file year for later deduplication.
all_extractions = []

for filepath in all_files:
    filename = os.path.basename(filepath)
    
    # Determining source file year and correct sheet.
    digits = ''.join(filter(str.isdigit, filename))
    if len(digits) < 4:
        continue
    file_year = int(digits[:4])
    
    if file_year < 2005:
        continue
    
    sheet = '6' if file_year <= 2014 else '7'
    
    try:
        extracted = extract_capex_from_file(filepath, sheet)
        for data_year, value in extracted.items():
            all_extractions.append({
                'Data_Year': data_year,
                'Source_File_Year': file_year,
                'Capex_100M_Yen': value
            })
        print(f"{filename}: extracted {len(extracted)} years — {list(extracted.keys())}")
    except Exception as e:
        print(f"{filename}: ERROR — {e}")

# Building the dataframe.
df_all = pd.DataFrame(all_extractions)
print(f"\nTotal rows before deduplication: {len(df_all)}")
print(df_all.head(10).to_string())
```

```python
def extract_dividends_from_file(filepath, sheet_name='4'):
    """
    Extracts year and Dividends values from MoF Sheet 4.
    Returns a dictionary of {year: dividends_value} in 100 million yen.
    """
    df = pd.read_excel(filepath, sheet_name=sheet_name, header=None)
    
    results = {}
    year_row_idx = None
    years = {}
    
    # Finding year row.
    for i in range(min(15, len(df))):
        row = df.iloc[i]
        year_cols = {}
        for j, val in enumerate(row):
            try:
                v = float(val)
                if 1998 <= v <= 2025 and v == int(v):
                    year_cols[j] = int(v)
            except (TypeError, ValueError):
                if isinstance(val, str):
                    cleaned = val.strip().replace('　', '')
                    try:
                        v = float(cleaned)
                        if 1998 <= v <= 2025 and v == int(v):
                            year_cols[j] = int(v)
                    except ValueError:
                        pass
        if len(year_cols) >= 2:
            year_row_idx = i
            years = year_cols
            break
    
    if not years:
        return {}
    
    # Finding dividends row.
    for i in range(year_row_idx, min(year_row_idx + 20, len(df))):
        for col_idx in range(min(3, len(df.columns))):
            cell = df.iloc[i, col_idx]
            if isinstance(cell, str) and 'Dividends' in cell:
                # Try extracting from same row first.
                for col, year in years.items():
                    val = df.iloc[i, col]
                    # If value is missing try next row.
                    if pd.isna(val) or not isinstance(val, (int, float)):
                        if i + 1 < len(df):
                            val = df.iloc[i + 1, col]
                    try:
                        v = float(val)
                        if v > 0 and not pd.isna(v):
                            results[year] = v
                    except (TypeError, ValueError):
                        pass
                if results:
                    return results
    
    return results

# Running extraction across all files.
all_dividends = []

for filepath in all_files:
    filename = os.path.basename(filepath)
    digits = ''.join(filter(str.isdigit, filename))
    if len(digits) < 4:
        continue
    file_year = int(digits[:4])
    if file_year < 2005:
        continue
    
    try:
        extracted = extract_dividends_from_file(filepath, '4')
        for data_year, value in extracted.items():
            all_dividends.append({
                'Data_Year': data_year,
                'Source_File_Year': file_year,
                'Dividends_100M_Yen': value
            })
        print(f"{filename}: extracted {len(extracted)} years — {list(extracted.keys())}")
    except Exception as e:
        print(f"{filename}: ERROR — {e}")

df_dividends = pd.DataFrame(all_dividends)
print(f"\nTotal rows before deduplication: {len(df_dividends)}")
print(f"\nUnique years: {sorted(df_dividends['Data_Year'].unique().tolist())}")
print(f"\nMissing years from 2003-2024: {[y for y in range(2003, 2025) if y not in df_dividends['Data_Year'].unique()]}")
```

```python
# Deduplicate rows to keep most recently published value per year.
df_dividends = df_dividends.sort_values(['Data_Year', 'Source_File_Year'])
dividends_final = df_dividends.drop_duplicates(
    subset='Data_Year', keep='last').copy()
dividends_final = dividends_final.sort_values(
    'Data_Year').reset_index(drop=True)

# Adding trillion yen column for readability.
dividends_final['Dividends_Trillion_Yen'] = (
    dividends_final['Dividends_100M_Yen'] / 10000).round(4)

print("Final dividends time series after deduplication:")
print(f"Shape: {dividends_final.shape}")
print()
print(dividends_final[['Data_Year', 'Source_File_Year',
                        'Dividends_100M_Yen',
                        'Dividends_Trillion_Yen']].to_string())

print(f"\nDividends range: {dividends_final['Dividends_Trillion_Yen'].min():.2f}T "
      f"to {dividends_final['Dividends_Trillion_Yen'].max():.2f}T yen")
print(f"Peak year: {dividends_final.loc[dividends_final['Dividends_100M_Yen'].idxmax(), 'Data_Year']}")
print(f"Trough year: {dividends_final.loc[dividends_final['Dividends_100M_Yen'].idxmin(), 'Data_Year']}")

# Saving.
dividends_final.to_csv("data/cleaned/mof_dividends_cleaned.csv", index=False)
print("\nDividends data saved successfully.")
```

```python
def extract_surplus_from_file(filepath, sheet_name='5'):
    """
    Extracts year and All Industries earned surplus values from MoF Sheet 5.
    Returns a dictionary of {year: surplus_value} in 100 million yen.
    Available from FY2015 onwards only.
    """
    df = pd.read_excel(filepath, sheet_name=sheet_name, header=None)
    
    results = {}
    year_row_idx = None
    years = {}
    
    # Finding year row.
    for i in range(min(10, len(df))):
        row = df.iloc[i]
        year_cols = {}
        for j, val in enumerate(row):
            try:
                v = float(val)
                if 1998 <= v <= 2025 and v == int(v):
                    year_cols[j] = int(v)
            except (TypeError, ValueError):
                if isinstance(val, str):
                    cleaned = val.strip().replace('　', '')
                    try:
                        v = float(cleaned)
                        if 1998 <= v <= 2025 and v == int(v):
                            year_cols[j] = int(v)
                    except ValueError:
                        pass
        if len(year_cols) >= 2:
            year_row_idx = i
            years = year_cols
            break
    
    if not years:
        return {}
    
    # Finding All Industries row.
    # Based on examination it consistently appears 2 rows after year row.
    search_terms = ['All Industries', 'All industries', 'ALL INDUSTRIES']
    
    for i in range(year_row_idx, min(year_row_idx + 10, len(df))):
        for col_idx in range(min(3, len(df.columns))):
            cell = df.iloc[i, col_idx]
            if isinstance(cell, str) and any(
                    term in cell for term in search_terms):
                for col, year in years.items():
                    val = df.iloc[i, col]
                    try:
                        v = float(val)
                        if v > 0 and not pd.isna(v):
                            results[year] = v
                    except (TypeError, ValueError):
                        pass
                if results:
                    return results
    
    return results

# Running extraction for FY2015 through FY2024 only.
all_surplus = []

for filepath in all_files:
    filename = os.path.basename(filepath)
    digits = ''.join(filter(str.isdigit, filename))
    if len(digits) < 4:
        continue
    file_year = int(digits[:4])
    
    # Earned surplus is only available from FY2015 onwards.
    if file_year < 2015:
        continue
    
    try:
        extracted = extract_surplus_from_file(filepath, '5')
        for data_year, value in extracted.items():
            all_surplus.append({
                'Data_Year': data_year,
                'Source_File_Year': file_year,
                'Earned_Surplus_100M_Yen': value
            })
        print(f"{filename}: extracted {len(extracted)} years — "
              f"{list(extracted.keys())}")
    except Exception as e:
        print(f"{filename}: ERROR — {e}")

df_surplus = pd.DataFrame(all_surplus)
print(f"\nTotal rows before deduplication: {len(df_surplus)}")
print(f"\nUnique years: {sorted(df_surplus['Data_Year'].unique().tolist())}")
print(f"\nMissing years from 2011-2024: "
      f"{[y for y in range(2011, 2025) if y not in df_surplus['Data_Year'].unique()]}")
```

```python
# Deduplicating and saving earned surplus.
df_surplus = df_surplus.sort_values(['Data_Year', 'Source_File_Year'])
surplus_final = df_surplus.drop_duplicates(
    subset='Data_Year', keep='last').copy()
surplus_final = surplus_final.sort_values(
    'Data_Year').reset_index(drop=True)

# Adding trillion yen column.
surplus_final['Earned_Surplus_Trillion_Yen'] = (
    surplus_final['Earned_Surplus_100M_Yen'] / 10000).round(4)

print("Final earned surplus time series after deduplication:")
print(f"Shape: {surplus_final.shape}")
print()
print(surplus_final[['Data_Year', 'Source_File_Year',
                      'Earned_Surplus_100M_Yen',
                      'Earned_Surplus_Trillion_Yen']].to_string())

print(f"\nSurplus range: {surplus_final['Earned_Surplus_Trillion_Yen'].min():.2f}T "
      f"to {surplus_final['Earned_Surplus_Trillion_Yen'].max():.2f}T yen")
print(f"\nGrowth from 2011 to 2024: "
      f"{((surplus_final.iloc[-1]['Earned_Surplus_100M_Yen'] / surplus_final.iloc[0]['Earned_Surplus_100M_Yen']) - 1) * 100:.1f}%")

# Saving.
surplus_final.to_csv("data/cleaned/mof_surplus_cleaned.csv", index=False)
print("\nEarned surplus data saved successfully.")
```

```python
# Fiscal Year to Calendar Year Alignment.
# Japan's fiscal year runs April 1 to March 31 of the following year.
# (FY2024 = April 2024 through March 2025)
# FY2024 data reflects corporate activity predominantly in calendar year 2024.

# Loading all MoF cleaned files.
capex = pd.read_csv("data/cleaned/mof_capex_cleaned.csv")
dividends = pd.read_csv("data/cleaned/mof_dividends_cleaned.csv")
surplus = pd.read_csv("data/cleaned/mof_surplus_cleaned.csv")
wb = pd.read_csv("data/cleaned/worldbank_cleaned.csv")
topix = pd.read_csv("data/cleaned/topix_cleaned.csv")

# Renaming Data_Year to Calendar_Year for clarity in the merged dataset.
capex = capex.rename(columns={'Data_Year': 'Calendar_Year'})
dividends = dividends.rename(columns={'Data_Year': 'Calendar_Year'})
surplus = surplus.rename(columns={'Data_Year': 'Calendar_Year'})
topix = topix.rename(columns={'Year': 'Calendar_Year'})

# Confirming data alignment.
print("Calendar year ranges after alignment:")
print(f"Capex: {capex['Calendar_Year'].min()} - {capex['Calendar_Year'].max()}")
print(f"Dividends: {dividends['Calendar_Year'].min()} - {dividends['Calendar_Year'].max()}")
print(f"Surplus: {surplus['Calendar_Year'].min()} - {surplus['Calendar_Year'].max()}")
print(f"TOPIX: {topix['Calendar_Year'].min()} - {topix['Calendar_Year'].max()}")
print(f"World Bank: {wb['Year'].min()} - {wb['Year'].max()}")
print(f"\nOverlapping period for all datasets: "
      f"{max(capex['Calendar_Year'].min(), wb['Year'].min(), topix['Calendar_Year'].min())} "
      f"to "
      f"{min(capex['Calendar_Year'].max(), wb['Year'].max(), topix['Calendar_Year'].max())}")
```

```python
# Building Master Merged Dataset.

# Building Japan-specific MoF dataset.
# Merging CapEx, dividends, and surplus on Calendar_Year.
japan_mof = capex[['Calendar_Year', 'Capex_100M_Yen', 
                    'Capex_Trillion_Yen']].copy()

japan_mof = japan_mof.merge(
    dividends[['Calendar_Year', 'Dividends_100M_Yen', 
               'Dividends_Trillion_Yen']],
    on='Calendar_Year', how='left'
)

japan_mof = japan_mof.merge(
    surplus[['Calendar_Year', 'Earned_Surplus_100M_Yen',
             'Earned_Surplus_Trillion_Yen']],
    on='Calendar_Year', how='left'
)

# Adding TOPIX data.
japan_mof = japan_mof.merge(
    topix[['Calendar_Year', 'TOPIX_Year_End', 
           'TOPIX_Annual_High', 'TOPIX_Annual_Low']],
    on='Calendar_Year', how='left'
)

print("Japan MoF + TOPIX merged dataset:")
print(f"Shape: {japan_mof.shape}")
print()
print(japan_mof.to_string())
print(f"\nMissing values:")
print(japan_mof.isnull().sum())
```

```python
# Extracting Japan World Bank data and merging with MoF data.
wb_japan = wb[wb['Country'] == 'Japan'].copy()

# Pivoting so each indicator becomes its own column.
wb_japan_pivot = wb_japan.pivot_table(
    index='Year', 
    columns='Indicator', 
    values='Value'
).reset_index()

# Flattening column names.
wb_japan_pivot.columns.name = None
wb_japan_pivot = wb_japan_pivot.rename(columns={
    'Year': 'Calendar_Year',
    'GDP (constant 2015 US$)': 'GDP_Constant_2015_USD',
    'GDP per capita (constant 2015 US$)': 'GDP_Per_Capita_Constant_2015_USD',
    'Gross domestic savings (% of GDP)': 'Gross_Domestic_Savings_Pct_GDP',
    'Gross fixed capital formation (% of GDP)': 'Gross_Fixed_Capital_Formation_Pct_GDP'
})

print("Japan World Bank pivot:")
print(f"Shape: {wb_japan_pivot.shape}")
print(wb_japan_pivot.head().to_string())

# Merging with Japan MoF dataset.
japan_master = japan_mof.merge(
    wb_japan_pivot,
    on='Calendar_Year',
    how='left'
)

print(f"\nJapan master dataset shape: {japan_master.shape}")
print(f"\nColumns: {japan_master.columns.tolist()}")
print(f"\nMissing values:")
print(japan_master.isnull().sum())
```

```python
# Building peer comparison dataset by extracting World Bank data for all four countries.

peer_countries = ['Japan', 'United States', 'Germany', 'Korea, Rep.']

wb_peers = wb[wb['Country'].isin(peer_countries)].copy()

# Pivoting to wide format with country-indicator combinations as columns.
wb_peers_pivot = wb_peers.pivot_table(
    index='Year',
    columns=['Country', 'Indicator'],
    values='Value'
).reset_index()

# Flattening multi-level column names.
wb_peers_pivot.columns = [
    '_'.join(col).strip() if col[1] else col[0] 
    for col in wb_peers_pivot.columns
]

wb_peers_pivot = wb_peers_pivot.rename(columns={'Year_': 'Calendar_Year'})

print("Peer comparison dataset:")
print(f"Shape: {wb_peers_pivot.shape}")
print(f"\nColumns: {wb_peers_pivot.columns.tolist()}")
print(f"\nFirst 3 rows:")
print(wb_peers_pivot.head(3).to_string())
print(f"\nMissing values:")
print(wb_peers_pivot.isnull().sum())
```
